```kotlin
import kotlin.math.pow

class Solution {
    fun plusOne(digits: IntArray): IntArray {
        val n = digits.size
        when (digits[n-1]){
            9 -> {
                if (n < 2) {
                    return intArrayOf(1, 0)
                }
                for (i in n-2 downTo 0){
                    if (digits[i+1] >= 9){
                        digits[i+1] = 0
                    }
                    digits[i]++
                    if (digits[i]<=9)
                        break
                }
                if(digits[0] == 10){
                    digits[0] = 0

                    val result = IntArray(n+1)
                    for (i in 1..(n-1))
                        result[i] = digits[i-1]
                    result[0] = 1
                    return result

                }
            }
            else -> digits[n-1]++
        }
        return digits
    }
}
```
