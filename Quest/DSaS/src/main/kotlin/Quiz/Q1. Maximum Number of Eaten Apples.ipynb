There is a special kind of apple tree that grows apples every day for n days. On the ith day, the tree grows apples[i] apples that will rot after days[i] days, that is on day i + days[i] the apples will be rotten and cannot be eaten. On some days, the apple tree does not grow any apples, which are denoted by apples[i] == 0 and days[i] == 0.

You decided to eat at most one apple a day (to keep the doctors away). Note that you can keep eating after the first n days.

Given two integer arrays days and apples of length n, return the maximum number of apples you can eat.



Example 1:

Input: apples = [1,2,3,5,2], days = [3,2,1,4,2]
Output: 7
Explanation: You can eat 7 apples:
- On the first day, you eat an apple that grew on the first day.
- On the second day, you eat an apple that grew on the second day.
- On the third day, you eat an apple that grew on the second day. After this day, the apples that grew on the third day rot.
- On the fourth to the seventh days, you eat apples that grew on the fourth day.
Example 2:

Input: apples = [3,0,0,0,0,2], days = [3,0,0,0,0,2]
Output: 5
Explanation: You can eat 5 apples:
- On the first to the third day you eat apples that grew on the first day.
- Do nothing on the fouth and fifth days.
- On the sixth and seventh days you eat apples that grew on the sixth day.


Constraints:

n == apples.length == days.length
1 <= n <= 2 * 104
0 <= apples[i], days[i] <= 2 * 104
days[i] = 0 if and only if apples[i] = 0.

```kotlin
fun eatenApples(apples: IntArray, days: IntArray): Int {
    var zeroApples = 0
    var furthestDay = 0
    var maxFurthestDay = 0

    for (i in apples.indices) {
        if (apples[i] != 0) {
            furthestDay = min(apples[i], days[i]) + i
            if (furthestDay > maxFurthestDay)
                maxFurthestDay = furthestDay
        }
        else {
         if (i > furthestDay - 1) {
             zeroApples++
         }
        }
    }

    return maxFurthestDay - zeroApples
}
```

```kotlin
import java.util.PriorityQueue

fun eatenApples(apples: IntArray, days: IntArray): Int {
    val heap = PriorityQueue<IntArray>() { a, b ->
        a[1].compareTo(b[1])
    }
    var result = 0
    var top = intArrayOf()

    for(i in apples.indices) {
        while(heap.isNotEmpty() && heap.peek()[1] <= i )
            heap.poll()

        if (apples[i] > 0)
            heap.add(intArrayOf(apples[i], days[i]+i))

        if (heap.isNotEmpty()) {
            result++
            top = heap.poll()
            if (top[0]>1)
                heap.add(intArrayOf(top[0]-1,top[1]))
        }
    }

    var day = apples.size
    var amount = 0

    while(heap.isNotEmpty()) {
        top = heap.poll()

        if (top[1]>day) {
            amount = min(top[0], top[1]-day)
            result += amount
            day = min(top[1], day + amount)
        }
    }

    return result
}
```

```kotlin
fun eatenApples(apples: IntArray, days: IntArray): Int {
    var result = 0

    val heap = PriorityQueue<IntArray>(Comparator.comparingInt { it[1] })

    for (i in 0 until apples.size + days.max()) {
        if (i < apples.size && apples[i] != 0)
            heap.add(intArrayOf(apples[i], days[i] + i))

        while (heap.isNotEmpty() && heap.peek()[1] <= i)
            heap.poll()

        if (heap.isNotEmpty()) {
            result++
            heap.peek()[0]--
            if (heap.peek()[0] == 0)
                heap.poll()
        }
    }

    return result
}
```
