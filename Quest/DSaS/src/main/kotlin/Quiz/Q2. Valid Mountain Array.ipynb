```kotlin
class Solution {
    fun validMountainArray(arr: IntArray): Boolean {
        var peak = arr[0]
        var increasing = true

        for (i in 1..arr.size-1){
            if (arr[i]>peak)
                peak = arr[i]

            if (arr[i]==arr[i-1])
                return false

            when(increasing){
                true ->{
                    if (arr[i]<arr[i-1])
                        increasing = false
                }
                false -> {
                    if (arr[i]>arr[i-1])
                        return false
                }
            }
        }
        return (arr[0]!=peak && arr[arr.size-1]!=peak)
    }
}
```
