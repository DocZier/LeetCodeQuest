```kotlin
import java.util.Stack

val s = "cbacdcbc"

val stack = Stack<Int>()
val seen = IntArray(26)
val last = IntArray(26)

s.forEachIndexed { index, ch ->
    last[ch-'a'] = index
}

for (i in 0..(s.length-1)){

    val ch = s[i] - 'a'

    if (seen[ch]++>0) continue

    while (!stack.empty() && stack.peek() > ch && i < last[stack.peek()])
        seen[stack.pop()] = 0

    stack.push(ch)
}

stack.asSequence().map { i ->
    (i + 'a'.code).toChar()
}.joinToString(separator = "")
```

```kotlin
class Solution {
    fun removeDuplicateLetters(s: String): String {
        val stack = Stack<Int>()
        val seen = IntArray(26)
        val last = IntArray(26)

        s.forEachIndexed { index, ch ->
            last[ch-'a'] = index
        }

        for (i in 0..(s.length-1)){

            val ch = s[i] - 'a'

            if (seen[ch]++>0) continue

            while (!stack.empty() && stack.peek() > ch && i < last[stack.peek()])
                seen[stack.pop()] = 0

            stack.push(ch)
        }

        return stack.asSequence().map { i ->
            (i + 'a'.code).toChar()
        }.joinToString(separator = "")
    }
}
```
