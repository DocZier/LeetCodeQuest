Given the head of a singly linked list, reverse the list, and return the reversed list.



Example 1:


Input: head = [1,2,3,4,5]
Output: [5,4,3,2,1]
Example 2:


Input: head = [1,2]
Output: [2,1]
Example 3:

Input: head = []
Output: []


Constraints:

The number of nodes in the list is the range [0, 5000].
-5000 <= Node.val <= 5000

```kotlin
class ListNode(var `val`: Int) {
    var next: ListNode? = null
}
```

```kotlin
import java.util.Stack

fun reverseList(head: ListNode?): ListNode? {
    if (head == null || head.next == null) return head

    val stack = ArrayDeque<ListNode>()
    var curNode = head
    var emptyNode = head

    while (curNode != null) {
        emptyNode = curNode
        curNode = curNode.next

        emptyNode.next = null
        stack.addLast(emptyNode)
    }

    val resultHead = stack.removeLast()
    curNode = resultHead

    while (stack.isNotEmpty()) {
        curNode?.next = stack.removeLast()
        curNode = curNode?.next
    }

    return resultHead
}
```

```kotlin
fun reverseList(head: ListNode?): ListNode? {
    if (head == null || head.next == null) return head

    var cur = head
    var prev: ListNode? = null

    while (cur != null) {
        val tmp = cur.next

        cur.next = prev
        prev = cur
        cur = tmp
    }

    return prev
}
```
