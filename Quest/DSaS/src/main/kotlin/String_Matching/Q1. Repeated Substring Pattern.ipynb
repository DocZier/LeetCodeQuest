Given a string s, check if it can be constructed by taking a substring of it and appending multiple copies of the substring together.



Example 1:

Input: s = "abab"
Output: true
Explanation: It is the substring "ab" twice.
Example 2:

Input: s = "aba"
Output: false
Example 3:

Input: s = "abcabcabcabc"
Output: true
Explanation: It is the substring "abc" four times or the substring "abcabc" twice.


Constraints:

1 <= s.length <= 104
s consists of lowercase English letters.

```kotlin
fun repeatedSubstringPattern(s: String): Boolean {

    val result = StringBuilder()
    var i = 1
    result.append(s[0])

    while (i < s.length-result.length) {
        if (s.substring(i, i + result.length) != result.toString()) {
            if (i / result.length >= 2) {
                for (j in 0..i / result.length-1)
                    result.append(result.toString())
                while (result.length > s.length)
                    result.deleteCharAt(result.length-1)
            }
            result.append(s[i])
        }
        else {
            i+= result.length-1
        }
        i++
        println(result)
    }

    return i + result.length <= s.length && s.substring(i, i + result.length) == result.toString()
}
```

```kotlin
fun repeatedSubstringPattern(str: String): Boolean {
    val sb = StringBuilder()
    for (i in str.length / 2 downTo 1) {
        if (str.length % i == 0) {
            sb.clear()
            val subS = str.substring(0, i)
            for (j in 0..<str.length / i) {
                sb.append(subS)
            }
            if (sb.toString() == str) return true
        }
    }
    return false
}
```
