Given two strings a and b, return the minimum number of times you should repeat string a so that string b is a substring of it. If it is impossible for b​​​​​​ to be a substring of a after repeating it, return -1.

Notice: string "abc" repeated 0 times is "", repeated 1 time is "abc" and repeated 2 times is "abcabc".



Example 1:

Input: a = "abcd", b = "cdabcdab"
Output: 3
Explanation: We return 3 because by repeating a three times "abcdabcdabcd", b is a substring of it.
Example 2:

Input: a = "a", b = "aa"
Output: 2


Constraints:

1 <= a.length, b.length <= 104
a and b consist of lowercase English letters.



```kotlin
fun repeatedStringMatch(a: String, b: String): Int {

    if (b.length < 1) return 0

    var repeaterCount = b.length / a.length + if (b.length % a.length != 0) 1 else 0

    val repeatedString = StringBuilder(a)

    repeat(repeaterCount-1){
        repeatedString.append(a)
    }
    if (repeatedString.contains(b)) return repeaterCount

    repeatedString.append(a)
    if (repeatedString.contains(b)) return ++repeaterCount

    return -1
}
```

Наиболее лучше данная задача решается через алгоритм Рабина-Карпа

```kotlin
fun repeatedStringMatch(a: String, b: String): Int {
    var result = 1
    if (b == a) return 1

    val frequency = IntArray(26)

    a.forEach { ch ->
        frequency[ch-'a']++
    }

    b.forEach { ch ->
        if (frequency[ch - 'a'] == 0) return -1
    }

    val repeatedString = StringBuilder(a)

    if (repeatedString.length < b.length) {
        result = b.length / repeatedString.length + if (b.length % a.length != 0) 1 else 0
        repeat(result-1) {
            repeatedString.append(a)
        }
    }

    if (repeatedString.indexOf(b) >= 0) return result
    repeatedString.append(a)
    if (repeatedString.indexOf(b) >= 0) return ++result
    return -1
}
```
