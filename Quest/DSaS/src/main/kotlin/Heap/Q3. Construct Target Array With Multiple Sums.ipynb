```kotlin
import java.util.PriorityQueue

fun isPossible(target: IntArray): Boolean {
    val heap = PriorityQueue<Int>() { a, b ->
        b.compareTo(a)
    }
    var totalSum = 0

    for (i in target) {
        heap.add(i)
        totalSum += i
    }

    while ( heap.peek() != 1 && target.size>1){
        var top = heap.poll()
        totalSum -= top

        if (top / totalSum < 1 ||  (top % totalSum == 0 && totalSum!=1)) {
            return false
        }

        heap.add(if (totalSum != 1) top % totalSum else 1 )
        totalSum += top % totalSum
    }

    return heap.peek() == 1
}
```

```kotlin
isPossible (intArrayOf(9,3,5))
```
