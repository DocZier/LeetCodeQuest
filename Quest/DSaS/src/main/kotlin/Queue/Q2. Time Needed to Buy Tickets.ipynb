```kotlin
class Solution {
    fun timeRequiredToBuy(tickets: IntArray, k: Int): Int {
        val queue = ArrayDeque<Int>()

        tickets.forEach {
            queue.add(it)
        }

        var timer = 0
        var pointer = k
        while (queue.isNotEmpty() && pointer!=-1){
            timer++
            queue[0]--
            if (pointer == 0) {
                if( queue[0]==0)
                    return timer
                else {
                    pointer = queue.size-1
                }
            }
            else
                pointer--
            if(queue[0] != 0)
                queue.addLast(queue[0])
            queue.removeFirst()

        }
        return timer
    }
}
```

```kotlin
class Solution {
    fun timeRequiredToBuy(tickets: IntArray, k: Int): Int {
        var timer = 0

        for (i in tickets.indices) {
            if(i>k)
                timer += min(tickets[i], tickets[k]-1)
            else
                timer += min(tickets[i], tickets[k])
        }

        return timer
    }
}
```
