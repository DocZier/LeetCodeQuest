```kotlin
import java.util.Stack

class MyQueue() {

    val stack = Stack<Int>()
    val backStack = Stack<Int>()
    val tmpStack = Stack<Int>()

    fun push(x: Int) {
        stack.push(x)

        while (!backStack.empty()) backStack.pop()

        while (!stack.empty()){
            tmpStack.push(stack.pop())
            backStack.push(tmpStack.peek())
        }

        while (!tmpStack.empty())
            stack.push(tmpStack.pop())

    }

    fun pop(): Int {
        val e = backStack.pop()

        while (!stack.empty()) stack.pop()

        while (!backStack.empty()){
            tmpStack.push(backStack.pop())
            stack.push(tmpStack.peek())
        }

        while (!tmpStack.empty())
            backStack.push(tmpStack.pop())

        return e
    }

    fun peek(): Int = backStack.peek()

    fun empty(): Boolean = backStack.empty()

}

/**
 * Your MyQueue object will be instantiated and called as such:
 * var obj = MyQueue()
 * obj.push(x)
 * var param_2 = obj.pop()
 * var param_3 = obj.peek()
 * var param_4 = obj.empty()
 */
```

```kotlin
import java.util.Stack

class MyQueue() {

    val inputStack = Stack<Int>()
    val outputStack = Stack<Int>()

    fun push(x: Int) = inputStack.push(x)

    fun pop(): Int {
        peek()
        return outputStack.pop()
    }

    fun peek(): Int {
        if (outputStack.empty())
            while (!inputStack.empty())
                outputStack.push(inputStack.pop())
        return outputStack.peek()
    }

    fun empty(): Boolean = inputStack.empty() && outputStack.empty()

}

/**
 * Your MyQueue object will be instantiated and called as such:
 * var obj = MyQueue()
 * obj.push(x)
 * var param_2 = obj.pop()
 * var param_3 = obj.peek()
 * var param_4 = obj.empty()
 */
```
