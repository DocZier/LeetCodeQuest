```kotlin
import java.util.Queue

class Solution {
    fun countStudents(students: IntArray, sandwiches: IntArray): Int {
        val studentQueue = ArrayDeque<Int>()
        var squareStudent = 0
        var squareSandwiches = 0

        for (i in students.indices){
            studentQueue.add(students[i])
            if (students[i]==1)
                squareStudent++
            if (sandwiches[i]==1)
                squareSandwiches++
        }

        var circleStudents = students.size - squareStudent
        var circleSandwiches = sandwiches.size - squareSandwiches
        var j = 0

        while (studentQueue.isNotEmpty() && j < sandwiches.size){
            println("Index ${j}: SquareSw: ${squareSandwiches}, CircleSw: ${circleSandwiches}| SquareSt: ${squareStudent}, CircleSt: ${circleStudents}")
            println("Students: ${studentQueue}, Sandwiches: ${sandwiches.toList()}")

            when (sandwiches[j]) {
                1 -> {
                    if (squareStudent==0) return circleStudents
                }
                0 -> {
                    if (circleSandwiches==0) return squareStudent
                }
            }

            if (sandwiches[j] == studentQueue[0]) {
                println("Index ${j}: SquareSw: ${squareSandwiches}, CircleSw: ${circleSandwiches}| SquareSt: ${squareStudent}, CircleSt: ${circleStudents}")
                when (sandwiches[j++]){
                    1 -> {
                        squareSandwiches--
                        squareStudent--
                    }
                    0 -> {
                        circleSandwiches--
                        circleStudents--
                    }
                }
            }
            else
                studentQueue.addLast(studentQueue[0])
            studentQueue.removeFirst()

        }

        return studentQueue.size
    }
}
```
