# Q3. Find all numbers disappeared in an array

Given an array nums of n integers where nums[i] is in the range [1, n], return an array of all the integers in the range [1, n] that do not appear in nums.



Example 1:

Input: nums = [4,3,2,7,8,2,3,1]
Output: [5,6]
Example 2:

Input: nums = [1,1]
Output: [2]

Самая простая идея: создать счётчик размерностью n. После прохода по массиву просто искать те, у которых значение ноль. В теории уложится можно спокойно в O(n)

```kotlin
class Solution {
    fun findDisappearedNumbers(nums: IntArray): List<Int> {

        val n = nums.size
        val counter = IntArray(n)
        val result = mutableListOf<Int>()

        for (i in 0..<n) {
            counter[nums[i]-1] += 1
        }

        for (i in 0..<n){
            if (counter[i]==0)
                result.add(i+1)
        }

        return result.toList()
    }
}
```

```kotlin
val solution = Solution()

val example1 = solution.findDisappearedNumbers(intArrayOf(4,3,2,7,8,2,3,1))

println(example1)

val example2 = solution.findDisappearedNumbers(intArrayOf(1,1))

println(example2)
```
