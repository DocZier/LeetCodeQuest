We define the usage of capitals in a word to be right when one of the following cases holds:

All letters in this word are capitals, like "USA".
All letters in this word are not capitals, like "leetcode".
Only the first letter in this word is capital, like "Google".
Given a string word, return true if the usage of capitals in it is right.



Example 1:

Input: word = "USA"
Output: true
Example 2:

Input: word = "FlaG"
Output: false


Constraints:

1 <= word.length <= 100
word consists of lowercase and uppercase English letters.

```kotlin
fun detectCapitalUse(word: String): Boolean {
    val count = word.count {it.code in 65..90}
    return count == word.length || count == 0 || (count == 1 && (word[0].code in 65..90))
}
```

```kotlin
fun detectCapitalUse(word: String): Boolean {
    when {
        word[0].code in 65..90  && word[word.length-1].code in 65..90-> {
            for (i in 1..word.length/2-1) {
                if (word[i].code !in 65..90 || word[word.length-1-i].code !in 65..90)
                    return false
            }
            return true
        }
        word[0].code in 65..90 && word[word.length-1].code !in 65..90 -> {
            for (i in 1..word.length/2-1) {
                if (word[i].code in 65..90 || word[word.length-1-i].code in 65..90)
                    return false
            }
            return true
        }
        else -> {
            for (i in 0..word.length/2) {
                if (word[i].code in 65..90 || word[word.length-1-i].code in 65..90)
                    return false
            }
            return true
        }
    }
}
```
