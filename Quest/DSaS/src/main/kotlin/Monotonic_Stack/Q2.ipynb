# Q2. Daily Temperatures

Given an array of integers temperatures represents the daily temperatures, return an array answer such that answer[i] is the number of days you have to wait after the ith day to get a warmer temperature. If there is no future day for which this is possible, keep answer[i] == 0 instead.


Example 1:

Input: temperatures = [73,74,75,71,69,72,76,73]
Output: [1,1,4,2,1,1,0,0]
Example 2:

Input: temperatures = [30,40,50,60]
Output: [1,1,1,0]
Example 3:

Input: temperatures = [30,60,90]
Output: [1,1,0]

```kotlin
class Solution {
    fun dailyTemperatures(temperatures: IntArray): IntArray {
        val result = IntArray(temperatures.size)

        for (i in 0..<temperatures.size){
            for (j in i+1..<temperatures.size){
                if(temperatures[j] > temperatures[i]){
                    result[i] = j - i
                    break
                }
            }
        }

        return result
    }
}
```

```kotlin
import java.util.Stack

class Solution {
    fun dailyTemperatures(temperatures: IntArray): IntArray {
        val result = IntArray(temperatures.size)

        val stack = Stack<Int>()

        for (i in temperatures.size-1 downTo 0){

            while (!stack.empty() && temperatures[stack.last()] <= temperatures[i] )
                stack.pop()

            if (!stack.empty())
                result[i] = stack.last() - i

            stack.push(i)
        }

        return result
    }
}
```

Идея такова:
Берём с конца добавляем в стек id элемент.Если элемент меньше или равен новому - удаляем, иначе -добавляем в стек. Для нахождения дистанции нам просто необходимо пройтись циклом,убирая верхушку из стека, пока не набредём на необходимое число, либо пока не пуст
